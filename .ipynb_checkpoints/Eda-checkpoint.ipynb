```python
numeric_columns = ['sy_snum', 'sy_pnum','pl_orbper',
       'pl_orbpererr1', 'pl_orbpererr2', 'pl_orbperlim', 'pl_orbsmax',
       'pl_orbsmaxerr1', 'pl_orbsmaxerr2', 'pl_orbsmaxlim', 'pl_rade',
       'pl_radeerr1', 'pl_radeerr2', 'pl_radelim', 'pl_bmasse',
       'pl_bmasseerr1', 'pl_bmasseerr2', 'pl_bmasselim', 'pl_orbeccen',
       'pl_orbeccenerr1', 'pl_orbeccenerr2', 'pl_orbeccenlim', 'pl_insol',
       'pl_insolerr1', 'pl_insolerr2', 'pl_insollim', 'pl_eqt', 'pl_eqterr1',
       'pl_eqterr2', 'pl_eqtlim', 'st_teff', 'st_tefferr1', 'st_tefferr2',
       'st_tefflim', 'st_rad', 'st_raderr1', 'st_raderr2', 'st_radlim',
       'st_mass', 'st_masserr1', 'st_masserr2', 'st_masslim', 'sy_dist',
       'sy_disterr1', 'sy_disterr2']
for col in numeric_columns:
    plt.figure(figsize = (6,4))
    sns.histplot(df[col],kde = True, bins = 20)
    plt.title(col)
    plt.show()
```

```python
for col in numeric_columns:
    plt.figure(figsize = (6,4))
    sns.histplot(np.log1p(df[col]), kde=True, bins=20)
    plt.title(col)
    plt.show()
```

```python
cols_to_drop = [col for col in df.columns if col.endswith(('err1', 'err2', 'lim'))]
df = df.drop(columns = cols_to_drop)
df.columns
```

```python
df.isnull().sum()
```

```python
df['pl_rade'].isnull().sum()
```

```python
df = df.dropna(subset = ['pl_rade'])
```

```python
df['pl_rade'].isnull().sum()
```

```python
df.shape
```

```python
def classify_planet(radius):
    if radius < 1.5:
        return 'rocky'
    elif radius < 4:
        return 'Super-Earth'
    elif radius < 10:
        return 'Neptune-Like'
    else:
        return 'Gas Giant'

df['planet_type'] = df['pl_rade'].apply(classify_planet)
```

```python
df['planet_type'].value_counts()
```
